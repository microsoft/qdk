```python
from qdk.estimator import LogicalCounts
from qdk.qre import estimate
from qdk.qre.application import QSharpApplication
from qdk.qre.models import GateBased, RoundBasedFactory, SurfaceCode

app = QSharpApplication(LogicalCounts({"numQubits": 100, "tCount": 10000, "cczCount": 1000, "measurementCount": 1000}))
architecture = GateBased(gate_time=50, measurement_time=1000)
architecture.usd_cost_per_hour = 3_600.0 * 1000

results = estimate(
    app,
    architecture,
    SurfaceCode.q() * RoundBasedFactory.q(),
    max_error=0.01,
)
results.as_frame()
```
