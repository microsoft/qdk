```python
import qodec as qc
import qdk.ec as ec

carbon_code = qc.Code(
    "carbon",
    stabilizers=[
        'X_0 X_1 X_2 X_3',
        'X_4 X_5 X_6 X_7',
        'X_8 X_9 X_10 X_11',
        'Z_0 Z_1 Z_2 Z_3',
        'Z_4 Z_5 Z_6 Z_7',
        'Z_8 Z_9 Z_10 Z_11',
        'X_0 X_1 X_5 X_7 X_8 X_11',
        'X_0 X_3 X_4 X_5 X_9 X_11',
        'Z_0 Z_2 Z_6 Z_7 Z_8 Z_11',
        'Z_0 Z_3 Z_4 Z_6 Z_10 Z_11'
    ],
    x=['X_0 X_2 X_4 X_5',  'X_0 X_1 X_5 X_6'],
    z=['Z_0 Z_1 Z_8 Z_11', 'Z_0 Z_2 Z_8 Z_9'],
)

carbon = ec.build_qodec(carbon_code)
print(carbon.summary())
```

```python
import qdk.qsharp
from qdk.simulation import NoiseConfig, run_qir
from collections import Counter

qdk.qsharp.init(target_profile=qdk.TargetProfile.Adaptive)
qir = qdk.qsharp.compile("{ use q = Qubit(); X(q); MResetZ(q) }")

noise = NoiseConfig()
noise.x.x = 0.01

Counter(run_qir(qir, shots=4_000, type="clifford", noise=noise, qodec=carbon))
```
