```python
import qdk
from qdk import qsharp
from qdk.simulation import run_qir
from collections import Counter

qsharp.init(target_profile=qdk.TargetProfile.Adaptive)
qir = qsharp.compile("""
{
    use q = Qubit();
    X(q);
    MResetZ(q)
}
""")
```

```python
# Circuit: X(q); MResetZ(q)

# At some point we could only run noiseless simulations
Counter(run_qir(qir, shots=4_000, type="clifford"))
```

```python
# Circuit: X(q); MResetZ(q)

# Currently, we can configure noise
from qdk.simulation import NoiseConfig

noise = NoiseConfig()
noise.x.x = 0.01
Counter(run_qir(qir, shots=4_000, type="clifford", noise=noise))
```

```python
# Now we can incorporate an error correction strategy.
from qdk.ec import develop

c4 = develop.load("c4.qodec.yaml")
Counter(run_qir(qir, shots=4_000, type="clifford", noise=noise, qodec=c4))
                                                                # New!
```
