```python
import qdk.ec
import qodec

steane_code = qodec.Code(
    "steane",
    stabilizers=[
        "X_0 X_3 X_4 X_6",
        "X_1 X_3 X_5 X_6",
        "X_2 X_4 X_5 X_6",
        "Z_0 Z_3 Z_4 Z_6",
        "Z_1 Z_3 Z_5 Z_6",
        "Z_2 Z_4 Z_5 Z_6",
    ],
    x=["X_0 X_1 X_3"],
    z=["Z_1 Z_2 Z_5"],
)

steane = qdk.ec.qodec_from_code(steane_code)
print(steane.summary())
```

```python
import qdk
from qdk import qsharp
from qdk.simulation import NoiseConfig, run_qir
from collections import Counter

qsharp.init(target_profile=qdk.TargetProfile.Adaptive)
qir = qsharp.compile("{ use q = Qubit(); X(q); MResetZ(q) }")

noise = NoiseConfig()
noise.x.x = 0.01

Counter(run_qir(qir, shots=4_000, type="clifford", noise=noise, qodec=steane))
```
