# Non-Clifford Stim extensions

The gate syntax is inspired by [Clifft's](https://github.com/unitaryfoundation/clifft/blob/main/docs/reference/gates.md). `qdk.stim` supports:

- **Phase gates:** `T`, `T_DAG`, `TPP`, `TPP_DAG`

- **Controlled gates:** `CH`, `CCX`, `CCZ`

- **Single-qubit rotations:** `R_X`, `R_Y`, `R_Z`, `U3`, `U`

- **Pair rotations:** `R_XX`, `R_YY`, `R_ZZ`

- **Pauli-product rotations:** `R_PAULI`
  

**Angles:** Unitless values are half turns: `R_X(0.5)` means $\pi/2$ radians. Append `rad` to specify radians directly: `R_X(0.5rad)`. `U3` arguments may mix units.

**Pauli products:** `!` negates a product, repeated same-qubit factors are folded, and anti-Hermitian products are rejected.

**Broadcasting:** Targets are grouped as qubits, consecutive pairs or triples, or whitespace-separated Pauli products, depending on the instruction.

**Simulation:** `type="clifford"` efficiently simulates Clifford circuits with a modest amount of non-Clifford computation, such as `T` gates and Pauli rotations. To achieve that, it uses stabilizer decomposition. For more general circuits, use `type="cpu"` or `type="gpu"`.

```python
from qdk import stim
from qdk.widgets import Histogram
```

## Phase gates

`T_DAG` reverses `T`, so applying both leaves qubit 0 unchanged. Four `T` gates equal `Z`; on qubit 1, the surrounding Hadamard gates turn that phase flip into a bit flip. `TPP_DAG` similarly reverses `TPP` on a Pauli product.

```python
phase_gates = """H 0 1
T 0
T_DAG 0
T 1
T 1
T 1
T 1
TPP X2*Z3
TPP_DAG X2*Z3
H 0 1
M 0 1 2 3
"""

Histogram(stim.run(phase_gates, shots=2000, type="clifford"), labels="kets")
```

## Controlled gates

`CH` applies a Hadamard when its control is `1`. `CCX` flips its target, and `CCZ` applies a phase flip, when both controls are `1`.

```python
controlled_gates = """X 0 1
CH 0 2
CCX 0 1 3
H 4
CCZ 0 1 4
H 4
M 0 1 2 3 4
"""

Histogram(stim.run(controlled_gates, shots=2000, type="clifford"), labels="kets")
```

## Rotation gates

`R_X` and `R_Y` rotate individual qubits, `R_ZZ` rotates a pair, `R_PAULI` rotates about a Pauli product, and `U` applies a general three-angle rotation,. Here, `U(1, 0, 0)` flips qubit 6. The examples use both half turns and radians.

```python
rotation_gates = """R_X(0.5) 0
R_Y(0.5rad) 1
R_ZZ(0.25) 2 3
R_PAULI(0.25) X4*Z5
U(1, 0, 0) 6
M 0 1 2 3 4 5 6
"""

Histogram(stim.run(rotation_gates, shots=2000, type="clifford"), labels="kets")
```
