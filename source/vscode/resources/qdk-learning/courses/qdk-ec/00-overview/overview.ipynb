## Why End-to-End QEC Tooling Matters

Developing a fault-tolerant quantum error correction protocol requires a QEC researcher to iterate across three tightly connected activities:
1. **Defining and Profiling Stabilizer Codes:** Choosing stabilizer generators, logical Pauli operators, and $[[n, k, d]]$ parameters that deliver the target encoding rate and distance under realistic physical noise models.
2. **Designing and Auditing Fault-Tolerant Gadgets:** Constructing circuits for state preparation, syndrome extraction, logical operations, and measurement that respect hardware connectivity constraints, specify parity checks and readouts, and prevent single circuit faults from spreading into undetected multi-qubit data errors (**problematic errors**).
3. **Compiling, Simulating, and Decoding Logical Experiments:** Compiling logical programs (written in **Q#** or **OpenQASM 3**) into QIR, simulating them under circuit-level noise, and decoding parity checks to evaluate logical error rates and rejection trade-offs.

Historically, QEC researchers had to stitch together incompatible ad-hoc scripts for stabilizer algebra, Stim circuit generation, MIP distance solving, QIR compilation, and decoding. **QDK-EC** unifies this entire workflow around a single open protocol representation (`qodec`).

#### The Magne Ecosystem: Logical Qubit Mode vs. Physical Qubit Mode
In the **Magne** neutral-atom quantum computing ecosystem ($>1,200$ Ytterbium-171 $^{171}\text{Yb}$ physical qubits with optical-tweezer all-to-all reconfigurability, zone-global Raman pulses, and high-fidelity Rydberg entangling gates), researchers interact with the stack in two complementary modes:
- **Logical Qubit Mode (Application Users):** Runs end-to-end fault-tolerant algorithms using Microsoft's built-in Qubit Virtualization stack.
- **Physical Qubit Mode (QEC Researchers — This Course):** Gives QEC researchers direct access to the **QDK-EC toolset** to design, verify, compile, and benchmark custom error-correcting codes, fault-tolerant gadgets, and decoding strategies tailored to neutral-atom capabilities.

#### Definitions of QDK-EC Capabilities & Open-Source Packages
The **QDK-EC toolset** is organized as a modular suite of open-source packages (`pip install qdk[ec] qodec paulimer binar`):

| Package / Module | Role in the QDK-EC Toolset | Key Capabilities & Algorithms |
| :--- | :--- | :--- |
| **`qodec`** *(Open Protocol Model)* | Hierarchical data model & `.qodec` YAML format for QEC protocols (`MIT` license). | Represents stacked **Layers**, where each layer binds an **`InstructionSet`** (declaring logical operations via `Stabilize`, `Clifford`, `Observe`, and `Pauli` actions plus `flags`), **`Code`** definitions (stabilizer generators & logical Paulis), and **`Gadget`** implementations (physical circuits, classical `readouts`, and parity `checks`). |
| **`qdk.ec`** *(Analysis, Synthesis & Verification)* | Static verification, automated protocol synthesis, and fault-distance analysis over `qodec` objects. | • **`CodeProfile(code)`**: Computes exact or bounded code distance (including biased/restricted noise `errors="Z"` and observable-specific distances), syndromes (`syndrome_of`), and logical actions (`logical_effect_of`).<br>• **`build_qodec(code, strategy)`**: Automatically synthesizes a multi-layer protocol (such as `"bare-css/v1"`).<br>• **`GadgetProfile(gadget)`**: Performs outcome-complete simulation to unfold any gadget into its logical **circuit action** (`profile.action`), compares it against `profile.objective`, computes **circuit distance** (`profile.distance()`), and traces fault propagation (`profile.effects_of(faults)`).<br>• **`audit(protocol)` & `filled(gadget)`**: Audits an entire protocol for missing observables, invalid checks, or incomplete frames, and automatically synthesizes missing stabilizer `checks` and `readouts`. |
| **`paulimer`** *(Pauli & Clifford Engine)* | Rust-backed symplectic Pauli algebra and 5-mode stabilizer simulation engine (`MIT` license). | Powers QDK-EC with five specialized stabilizer simulators:<br>1. **Outcome-specific:** Traditional trajectory simulation with random measurement outcomes.<br>2. **Outcome-complete:** Simultaneously represents *all* possible measurement outcome branches in $O(n_{\text{qubits}}^2 + n_{\text{qubits}} n_{\text{outcomes}})$ time — enabling `GadgetProfile` to verify gadgets analytically.<br>3. **Outcome-free:** Tracks Clifford operations up to Pauli corrections without sampling.<br>4. **Faulty simulation:** High-throughput noisy Pauli frame sampling.<br>5. **Frame propagation:** Propagates Pauli error frames through Clifford circuits. |
| **`binar`** *(Binary Linear Algebra)* | Portable Rust $\mathbb{F}_2$ bit-matrix & affine-space library (`MIT` license). | SIMD-accelerated `BitMatrix` and `BitVector` operations, echelon reduction, kernel/nullspace calculation (`H.kernel()`), and linear system solving over $\mathbb{F}_2$ used across check synthesis and decoding. |
| **`qdk.qsharp` & `qdk.openqasm`** *(Logical Compilation)* | Compiles high-level logical quantum programs against a `qodec` layer into QIR. | Compiles **Q#** (`@Measurement() ... body intrinsic;`) and **OpenQASM 3** (`@qdk.qir.intrinsic gate`) with `TargetProfile.Adaptive_RI` so each high-level `Qubit` represents an encoded `qodec` block. |
| **`qdk.simulation` & `deq`** *(Noisy Simulation & Decoding)* | Circuit-level Monte Carlo execution (`run_qir`) and modular syndrome decoding. | Executes compiled QIR against a `qodec` protocol under customizable `NoiseConfig` models. **`deq`** compiles stabilizer checks and circuit error models just-in-time (`check_model="jit"`, `error_model="jit"`) and coordinates streaming/windowed or monolithic decoders (`black-box-relay-bp`, `pymatching`) with soft readout-score gap filtering (`max_readout_score`) for confidence post-selection. |

> **Current QDK-EC Preview Scope:**
> - **Supported today:** Stabilizer codes (`qodec.Code`), CSS/Clifford gadgets, outcome-complete circuit-action verification (`GadgetProfile`), static MIP/enumeration code & circuit distance, automatic check/readout synthesis (`ec.filled`), Q# and OpenQASM 3 logical compilation with static control branching, and single-gadget / monolithic `deq` & syndrome decoding under depolarizing and biased Pauli noise.
> - **Upcoming on the roadmap:** Dynamic classical control branching (`if (m == One)` branching across encoded blocks in `run_qir`), stateful multi-round sliding-window decoding across back-to-back gadgets, and non-Clifford magic-state injection gadgets.

<div style="padding:0.1em 1em;">

Within the QDK-EC toolset, `qodec` provides the open protocol data model and serialization format (`.qodec` YAML) for hierarchical fault-tolerant protocols—organizing Layers, InstructionSets, Codes, and Gadgets. The `qdk.ec` module provides the static analysis, synthesis, and verification algorithms that operate on `qodec` objects, such as `CodeProfile` (code distance and syndromes), `GadgetProfile` (circuit action and circuit distance), `build_qodec`, `audit`, and `filled`.

<div style="border-left:4px solid #8c4a00;background:#8c4a001a;border-radius:4px;margin:1em 0;"><details><summary style="background:#8c4a00;color:#ffffff;padding:0.35em 0.8em;cursor:pointer;font-weight:600;">&#10067;&nbsp;How does a <code style="color:inherit;font-family:var(--vscode-editor-font-family,monospace);background:rgba(0,0,0,0.2);border-radius:3px;padding:0 0.2em;">qodec</code> separate a logical instruction's specification from its circuit implementation, and how does <code style="color:inherit;font-family:var(--vscode-editor-font-family,monospace);background:rgba(0,0,0,0.2);border-radius:3px;padding:0 0.2em;">qdk.ec</code> verify that the two match?</summary>

<div style="padding:0.1em 1em;">

In a `qodec`, an `Instruction` in an `InstructionSet` declares the algebraic contract (`Stabilize`, `Clifford`, `Pauli`, or `Observe` actions and optional `flags`), while a `Gadget` provides the implementing circuit, block `in`/`out` encodings, parity `checks`, and `readouts`. `qdk.ec.GadgetProfile(gadget)` computes the **circuit action** (`profile.action`) from the gadget circuit, checks, and readouts and verifies that it matches the declared instruction (`profile.objective`).
