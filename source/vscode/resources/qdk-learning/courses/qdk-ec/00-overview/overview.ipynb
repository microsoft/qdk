## Why a formal protocol model matters

A working QEC experiment couples four distinct artifacts: a **stabilizer code**, the **logical operations** exposed to an algorithm, the **circuits** that implement those operations, and the **parity equations** that connect physical measurements to logical outcomes, flags, and syndromes. When those connections live implicitly inside circuit-generation scripts, editing a schedule or swapping a decoder forces you to reconstruct what every qubit and measurement bit means.

A **qodec** solves this problem by recording codes, instruction sets, and gadgets in one formal, serializable protocol model. Because each logical instruction declares its mathematical action separately from the circuit that implements it, tools can verify circuits automatically, analyze how faults propagate, and execute compiled programs against interchangeable protocols. Crucially, a circuit can implement the right logical operation without faults and still fail after one physical fault—closing that gap drives every step of this course.

## How the QDK error correction tools fit together

The QDK error correction workflow divides responsibilities across four components:

| Component | Primary responsibility |
| :--- | :--- |
| `qodec` | Defines and serializes codes, instruction sets, gadgets, layers, and complete QEC protocols. |
| `qdk.ec` | Analyzes codes and gadgets, builds baseline protocols, derives equations, and audits consistency. |
| `qdk.simulation` | Executes QIR programs through a `qodec` under configurable gate noise and decodes runs (built-in syndrome/frame decoders or `deq`). |
| `qdk.qre` | Estimates architectural space, time, and error overhead for resource estimation (separate from `qodec` execution). |

## End-to-end workflow preview

The units ahead unpack the three workflow stages from the diagram above into focused, testable steps:

- **1. Develop (`Units 01–03`):** Define and profile a stabilizer code (`qodec.Code`, `ec.CodeProfile`), build a two-layer baseline protocol (`ec.build_qodec`), and inspect its gadget circuits, checks, and readouts.
- **2. Test (`Units 04–06`):** Validate noiseless circuit actions and derive missing equations (`ec.filled`, `ec.audit`), identify circuit faults that lower gadget distance (`ec.GadgetProfile`), and upgrade gadgets for fault tolerance.
- **3. Deploy (`Units 07–10`):** Compile Q# and OpenQASM logical experiments to QIR, decode noisy shots with `qdk.simulation.run_qir` and `deq`, and evaluate a custom hardware-motivated gadget in the Capstone.

Run the preview cell below to see the core **Develop** and **Test** objects in action on the `C4` code:

```python
import qodec
from qdk import ec

c4 = qodec.Code(
    "C4",
    stabilizers=["X_0 X_1 X_2 X_3", "Z_0 Z_1 Z_2 Z_3"],
    x=["X_0 X_1", "X_0 X_2"],
    z=["Z_0 Z_2", "Z_0 Z_1"],
)

code_profile = ec.CodeProfile(c4)
print("Code distance:", code_profile.distance())

protocol = ec.build_qodec(c4, strategy="bare-css/v1", strict=False)
print(ec.audit(protocol))
```

Notice the final line: `audit: ok`. That confirms every circuit and equation in the baseline protocol implements its declared logical operation when **no faults occur**.

Static protocol validation verifies noiseless semantics only: *Analyzing Gadget Distance and Circuit Faults* shows that several baseline gadgets still have circuit-level distance `1` until their schedules and flags are upgraded.

<div style="padding:0.1em 1em;">

`ec.audit` checks noiseless consistency: whether the circuits, checks, readouts, and frames match their declared logical instructions when no faults happen. A circuit can act correctly without faults yet allow one physical gate fault to spread through an ancilla into an undetected multi-qubit logical error.
